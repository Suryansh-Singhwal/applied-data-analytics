# Descriptive Statistics

Descriptive statistics is a method for summarizing and describing the main features of a set of data. It helps us understand the basic details of the data, like:

1. Mean (Average): What the "typical" value is.
2. Median: The middle value when data is sorted.
3. Mode: The most frequent value.
4. Range: The difference between the highest and lowest values.
5. Standard Deviation: How spread out the data is from the average.

In short, descriptive statistics summarizes data and makes it easier to understand and interpret.

## Frequency Distribution

A frequency distribution is a way to organize data to show how often each value or group of values occurs.

It is usually presented as a table where:

- The first column contains values or groups of values.
- The second column contains their frequencies or counts.

### Example

The following dataset represents the daily demand for hammers at a hardware store over the past 20 days.

Develop a frequency distribution to summarize this data.

```python
# Importing required libraries

import pandas as pd
import numpy as np
```

```python
# Generating the dataset

daily_demand = [
    2, 1, 0, 2, 1, 3, 0, 2, 4, 0,
    3, 2, 3, 4, 2, 2, 2, 4, 3, 0
]

# Display the dataset
print(daily_demand)

# Check the data type
print(type(daily_demand))
```

```python
# Creating a pandas DataFrame

df1 = pd.DataFrame(
    daily_demand,
    columns=['DailyDemand']
)

# Display the first 5 records
df1.head()
```

The DataFrame contains one column named DailyDemand.

The head() function displays the first five records of the dataset.

```python
# Creating a frequency distribution

summary1 = (
    df1
    .groupby('DailyDemand')
    .size()
    .reset_index(name='Frequency')
    .sort_values(by='DailyDemand')
)

# Display the frequency distribution
summary1
```

The frequency distribution shows how many times each daily demand value occurred.

For example:

- Demand of 0 occurred 4 times.
- Demand of 1 occurred 2 times.
- Demand of 2 occurred 7 times.
- Demand of 3 occurred 4 times.
- Demand of 4 occurred 3 times.

## Relative Frequency

Relative frequency is the proportion or percentage of times a specific value occurs in a dataset compared to the total number of observations.

Formula:

Relative Frequency = Frequency / Total Frequency

Relative frequency can also be multiplied by 100 to represent it as a percentage.

```python
# Calculating relative frequency

summary1['RelativeFrequency'] = (
    summary1['Frequency'] /
    summary1['Frequency'].sum()
)

# Display the result
summary1
```

The sum of all relative frequencies should always equal 1.00, which represents 100%.

In this dataset:

- Demand 0 has a relative frequency of 0.20 or 20%.
- Demand 1 has a relative frequency of 0.10 or 10%.
- Demand 2 has a relative frequency of 0.35 or 35%.
- Demand 3 has a relative frequency of 0.20 or 20%.
- Demand 4 has a relative frequency of 0.15 or 15%.

## Cumulative Relative Frequency

Cumulative relative frequency is the running total of relative frequencies.

For each row:

Cumulative Relative Frequency =
Current Relative Frequency + Previous Cumulative Relative Frequency

The final cumulative relative frequency should always equal 1.00 or 100%.

```python
# Calculating cumulative relative frequency

summary1['CumulativeRelativeFrequency'] = (
    summary1['RelativeFrequency'].cumsum()
)

# Display the result
summary1
```

The cumulative relative frequency allows us to understand the proportion of observations up to a particular value.

For example, the cumulative relative frequency for a daily demand of 2 is 0.65.

This means that 65% of the observations had a daily demand of 2 or less.

## Frequency Distribution Using Classes or Bins

Consider another dataset representing the number of calls made per day from a cell phone over the past 30 days.

Develop a frequency distribution to summarize the data.

Because this dataset contains many possible values, instead of listing every individual value, we can group the values into intervals called classes or bins.

### Determining the Number of Classes

One common rule for determining the number of classes is:

2^k >= n

Where:

- k = number of classes
- n = number of observations

For this dataset:

n = 30

We can choose:

k = 5

because:

2^5 = 32

and:

32 >= 30

Therefore, 5 classes are appropriate for this dataset.

### Calculating Class Width

The class width can be calculated using:

Class Width = (Maximum Value - Minimum Value) / Number of Classes

The result is rounded up to the next whole number.

We will use the calculated class width to create the class intervals.

```python
# Generating the dataset

calls_per_day = [
    4, 5, 1, 0, 7, 8, 3, 6, 8, 3,
    0, 9, 2, 12, 14, 5, 5, 10, 7, 2,
    11, 9, 4, 3, 1, 5, 7, 3, 5, 6
]

# Creating a pandas DataFrame

df2 = pd.DataFrame(
    calls_per_day,
    columns=['CallsPerDay']
)

# Display the first 5 records
df2.head()
```

```python
# Determining the number of classes

k = 5

# Finding minimum and maximum values

min_value = df2['CallsPerDay'].min()
max_value = df2['CallsPerDay'].max()

# Calculating class width
# np.ceil() rounds the value up to the next whole number

w = np.ceil(
    (max_value - min_value) / k
).astype(int)

# Display class width
w
```

The calculated class width is 3.

Therefore, we will create class intervals with a width of 3.

```python
# Creating the class intervals

bins = np.arange(
    min_value,
    max_value + w,
    w
)

# Display the bins
bins
```

The class boundaries are:

0, 3, 6, 9, 12, 15

These boundaries create the following classes:

0–3

3–6

6–9

9–12

12–15

```python
# Assigning each observation to a class

df2['Class'] = pd.cut(
    df2['CallsPerDay'],
    bins=bins,
    include_lowest=True
)

# Creating the frequency distribution

summary2 = (
    df2
    .groupby(
        'Class',
        observed=True
    )['CallsPerDay']
    .count()
    .reset_index(name='Occurrence')
)

# Display the frequency distribution
summary2
```

```python
# Calculating relative frequency

summary2['RelativeFrequency'] = (
    summary2['Occurrence'] /
    summary2['Occurrence'].sum()
)

# Calculating cumulative relative frequency

summary2['CumulativeRelativeFrequency'] = (
    summary2['RelativeFrequency'].cumsum()
)

# Display the complete frequency distribution

summary2
```

## Interpretation

The final frequency distribution contains:

- Class interval
- Occurrence/Frequency
- Relative Frequency
- Cumulative Relative Frequency

For example:

The class (-0.001, 3.0] contains 10 observations.

Its relative frequency is approximately 0.3333, which means approximately 33.33% of all observations fall within this class.

The final cumulative relative frequency is 1.00, meaning that 100% of the observations have been included.

## Key Takeaways

1. Frequency tells us how many times a value occurs.
2. Relative frequency tells us the proportion of observations represented by a value.
3. Cumulative relative frequency gives the running total of relative frequencies.
4. Classes or bins are useful when a dataset contains many possible values.
5. The rule 2^k >= n can be used to determine an appropriate number of classes.
6. Class width determines the size of each interval.
7. The final cumulative relative frequency should always be 1.00 or 100%.
